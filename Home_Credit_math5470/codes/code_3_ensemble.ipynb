## Ensemble

```python
# libraries
import pandas as pd
import numpy as np
from scipy.stats import rankdata
```

```python
# import data
a = pd.read_csv("../submissions/xgb.csv")
b = pd.read_csv("../submissions/lgb.csv")
```

```python
# rank mean
a["TARGET"] = rankdata(a.TARGET) + rankdata(b.TARGET)
```

```python
# scaling
a["TARGET"] = (a.TARGET - a.TARGET.min()) / (a.TARGET.max() - a.TARGET.min())
```

```python
# export CSV
a.to_csv("../submissions/ensemble_subm.csv", index = False, float_format = "%.8f")
```
